```python
import matplotlib.pyplot as plt

def Picos(L):
    if len(L) < 3:
        return 0
    pico = 1 if L[0] < L[1] > L[2] else 0
    return pico + Picos(L[1:])

# Lista de ejemplo
L = [1,0,7,0,2,-23]

# Encontrar índices de picos
indices_picos = [i for i in range(1, len(L)-1) if L[i-1] < L[i] > L[i+1]]

# Graficar
plt.plot(L, marker='o', color='blue', label='Valores')
plt.scatter(indices_picos, [L[i] for i in indices_picos], color='red', label='Picos', zorder=5)
plt.title(f'Picos en la lista {L}')
plt.xlabel('Índice')
plt.ylabel('Valor')
plt.legend()
plt.grid(True)
plt.show()
```
