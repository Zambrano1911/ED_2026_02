```python
class StackPQ():
    def __init__(self):
        self.data = UnsortedPriorityQueue()
    
    def push(self, e):
        self.data.add = (self.key, e)
        self.key = 0

    def top(self):
        self.data.min()[1]
    
    def pop(self):
        if len(self.data) == 0:
            raise Empty('El stack está vacio')
        return self.data.remove_min()[1]
```
