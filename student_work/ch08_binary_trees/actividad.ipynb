```python
def find_min(T:LinkedBinaryTree):
    if T.is_empty():
        return None
    p = Troot()
    while T._left(p) is not None:
        min = T._left(p)
    return min.element()

def es_ordenado(T:LinkedBinaryTree):
    I = T.inorder()
    a = next(T)
    for i in range(len(T)-2):
        b = next(I)
        if b.element() < a.element():
            return False
        a = next(I)
        a, b = b, a
    return True
```
